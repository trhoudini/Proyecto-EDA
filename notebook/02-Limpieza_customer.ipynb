## Carga de datos y leer todas las hojas del archivo Excel

```python
import pandas as pd

# Ruta del archivo Excel
archivo_excel = '../data/raw/customer-details.xlsx'

# Leer todas las hojas en un diccionario
hojas = pd.read_excel(archivo_excel, sheet_name=None)

# Ver los nombres de las hojas
print("Hojas encontradas:", hojas.keys())
```

## Guardar cada hoja en su propio DataFrame

```python
df_2012 = hojas['2012']
df_2013 = hojas['2013']
df_2014 = hojas['2014']
```

## Vamos a empezar limpiar la hoja de 2012

```python
df_2012.head()
```

## Selección de columnas

Vamos a eliminar la columna de `Unnamed: 0` ya que no nos aporta nada de información. Se trata de un índice antiguo que se guardó por error al exportar el Excel. No tiene valor analítico y puede eliminarse sin problema.

## Eliminar duplicados y nulos

No hay duplicados ni nulos en el dataset.

## Eliminación de de la columna `Unnamed: 0`

```python
df_2012 = df_2012.drop_duplicates()
```

```python
columnas_eliminar = [
    'Unnamed: 0'
    ]
```

```python
df_2012.drop(columns=columnas_eliminar)
```

Ya hemos eliminado la columna `Unnamed: 0` de la hoja de 2012 del dataset, y cómo ya sabemos que no hay nulos ni duplicados en esta hoja, pasamos a la limpieza de la hoja del 2013.
